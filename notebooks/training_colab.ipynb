```python
import torch

print("=== CUDA / GPU ===")
print("PyTorch :", torch.__version__)
print("CUDA disponible :", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU :", torch.cuda.get_device_name(0))
    print("Version CUDA :", torch.version.cuda)

    total_memory = torch.cuda.get_device_properties(0).total_memory
    print("VRAM :", round(total_memory / 1024**3, 2), "GB")
else:
    print("⚠️ Aucun GPU CUDA disponible")
```

```python
import os
import subprocess

os.chdir("/content/llm-dataset")

# ==========================================
# Installation de Poetry
# ==========================================

subprocess.run(
    ["pip", "install", "poetry"],
    check=True
)

subprocess.run(
    ["poetry", "--version"],
    check=True
)

# ==========================================
# Configuration PyTorch CUDA pour Colab
# ==========================================

pyproject_path = "pyproject.toml"

with open(pyproject_path, "r", encoding="utf-8") as f:
    content = f.read()

content = content.replace(
    'url = "https://download.pytorch.org/whl/cpu"',
    'url = "https://download.pytorch.org/whl/cu126"'
)

with open(pyproject_path, "w", encoding="utf-8") as f:
    f.write(content)

print("Source PyTorch configurée pour CUDA 12.6.")

# ==========================================
# Régénération du lock Poetry
# ==========================================

subprocess.run(
    ["poetry", "lock"],
    check=True
)

# ==========================================
# Installation des dépendances avec Poetry
# ==========================================

subprocess.run(
    ["poetry", "install"],
    check=True
)

print("Installation Poetry terminée.")
```
