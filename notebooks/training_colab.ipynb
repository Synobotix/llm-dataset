```python

import os
import subprocess
from pathlib import Path

PROJECT = Path("/content/llm-dataset")
os.chdir(PROJECT)

# --------------------------------------------------
# 1. Installer Poetry dans Colab
# --------------------------------------------------

subprocess.run(
    ["pip", "install", "poetry"],
    check=True
)

subprocess.run(
    ["poetry", "--version"],
    check=True
)

# --------------------------------------------------
# 2. Modifier uniquement la source PyTorch
# --------------------------------------------------

pyproject_path = PROJECT / "pyproject.toml"

content = pyproject_path.read_text(encoding="utf-8")

content = content.replace(
    'url = "https://download.pytorch.org/whl/cpu"',
    'url = "https://download.pytorch.org/whl/cu126"'
)

pyproject_path.write_text(content, encoding="utf-8")

print("Source PyTorch configurée pour CUDA 12.6.")

# --------------------------------------------------
# 3. Supprimer le lock CPU cloné depuis GitHub
# --------------------------------------------------

lock_path = PROJECT / "poetry.lock"

if lock_path.exists():
    lock_path.unlink()
    print("Ancien poetry.lock supprimé.")
else:
    print("Aucun poetry.lock trouvé.")

# --------------------------------------------------
# 4. Créer le nouveau lock CUDA
# --------------------------------------------------

subprocess.run(
    ["poetry", "lock"],
    cwd=PROJECT,
    check=True
)

print("Nouveau poetry.lock généré.")

# --------------------------------------------------
# 5. Installer les dépendances avec Poetry
# --------------------------------------------------

subprocess.run(
    ["poetry", "install"],
    cwd=PROJECT,
    check=True
)

print("Installation Poetry terminée.")

```
